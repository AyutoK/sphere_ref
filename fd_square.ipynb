```python
import numpy as np
import ultraplot as  uplt
import matplotlib.pyplot as plt
import sphere_ref_lib as srl
import sphere_variables as sv
import pandas as pd
```

```python
fp = "./output_focus/"
fp_nc = srl.set_output_dir_nc()
R, step, xs, d, Z0, _ = srl.set_basic_params()
_, _, target, apply_ref, _, _ = sv.svv()
e1, e2, H_obs, D_moon, R_moon, e1, e2, tandelta = srl.get_default_param(target)
height = np.array([100,200,500,1000])

R2s =( height + (R_moon / 1000) )/ (R_moon/1000) # ganymedeにおいて100km, 200km, 500km, 1000kmを想定
ths = np.arange(0,91,1)
ths_rad = np.radians(ths)
focus_p = 0.5

plot_colors = ["red", "darkorange", "springgreen", "mediumblue"]

Rax, thax = np.meshgrid(R2s, ths_rad)
```

```python
Rf = np.sqrt(focus_p ** 2 + Rax ** 2 - 2 * focus_p * Rax * np.cos(thax))

Rfd = pd.DataFrame(Rf, index=ths, columns=height)

Rfd_power = 0.5 ** 2 / Rfd ** 2

Rfd_power
```

```python
cycle = uplt.Cycle(colors=plot_colors)

df_accu = pd.read_csv("./csv_highspeed/line_plot_distance_ratio.csv", index_col=0)
df_accu.columns = [str(x) + "km" for x in height]

df_accu
```

```python
cycle = uplt.Cycle(colors=plot_colors)

fig, ax = uplt.subplots(suptitle="Rfd_power", xlabel="theta (degree)", ylabel="Rfd_power", refwidth=8, refheight=5)
ax.format(ylim=(0,1.0), ylocator=0.1)
ax.plot(ths, Rfd_power, cycle=cycle, label=[f"{h} km 1/2 focus" for h in height], ls="--", legend="ur")
ax.plot(ths, df_accu.iloc[0:91,:], cycle=cycle, label=[f"{h} km accurate" for h in height], legend="ll")

uplt.show()
```

```python
left_col = df_accu.iloc[:, 0].replace(0, np.nan)
df_accu_ratio_sq = df_accu.div(left_col, axis=0) ** (-1/2)

df_accu_red = df_accu_ratio_sq * height[0]

df_accu_red
```

```python
fig, ax = uplt.subplots(suptitle="distance ratio from focus", xlabel="theta (degree)", ylabel="distance ratio", refwidth=8, refheight=5)
ax.format(xlim=(0,90))
ax.plot(df_accu_red, cycle=cycle, label=[f"{h} km" for h in height], legend="lr")

uplt.show()
```
