```python
import numpy as np
import ultraplot as  uplt
import matplotlib.pyplot as plt
import sphere_ref_lib as srl
import sphere_variables as sv
import pandas as pd
```

```python
fp = "./output_focus/"
fp_nc = srl.set_output_dir_nc()
R, step, xs, d, Z0, _ = srl.set_basic_params()
_, _, target, apply_ref, _, _ = sv.svv()
e1, e2, H_obs, D_moon, R_moon, tandelta = srl.get_default_param(target)
height = np.array([1,100,200,500,1000])

R2s =( height + (R_moon / 1000) )/ (R_moon/1000) # ganymedeにおいて100km, 200km, 500km, 1000kmを想定
ths = np.arange(0,91,1) #入射角(0-90deg)
ths_rad = np.radians(ths)


plot_colors = ["red", "darkorange", "springgreen", "mediumblue", "fuchsia"]

Rax, thax = np.meshgrid(R2s, ths_rad)

R2s
```

```python
# 入射角から探査機角度を導出
# alpha_rad = srl.calc_alpha(thax,(Rax-1),1)
# ald = pd.DataFrame(np.degrees(alpha_rad), index=ths, columns=[str(x) + "km" for x in height])
# ald

alpha_deg = srl.calc_alpha(np.degrees(thax),(Rax-1),1)
ald = pd.DataFrame(alpha_deg, index=ths, columns=[str(x) + "km" for x in height])
alpha_rad = np.radians(alpha_deg)
ald
```

```python
focus_p = 0.5

Rf = np.sqrt(focus_p ** 2 + Rax ** 2 - 2 * focus_p * Rax * np.cos(alpha_rad))

Rfd = pd.DataFrame(Rf, index=ths, columns=[str(x) + "km" for x in height])
Rfd_norm = np.sqrt(1 ** 2 + focus_p ** 2 - np.cos(thax))

Rfd_power = Rfd_norm ** 2 / Rfd ** 2

Rfd_power
```

```python
cycle = uplt.Cycle(colors=plot_colors)

df_accu = pd.read_csv("./csv_highspeed/line_plot_distance_ratio.csv", index_col=0)
df_accu.columns = [str(x) + "km" for x in height]
df_accu_ax = df_accu.index.astype(float)
df_accu_ax
```

```python
fig, ax = uplt.subplots(suptitle="Rfd_power fixed focus point", xlabel="s/c angle alpha (degree)", ylabel="Rfd_power", figsize=(10,6))
ax.format(ylim=(0,1.0), ylocator=0.1)
ax.plot(ald, Rfd_power, cycle=cycle, label=[f"{h} km 1/2 focus" for h in height], ls="--", legend="ur")
ax.plot(df_accu_ax, df_accu, cycle=cycle, label=[f"{h} km accurate" for h in height], legend="ll")

fig.savefig(fp + "Rfd_power_fixed_focus.png")
fig.show()
```

```python
focus_p = 1/(2 * np.cos(thax)) 

Rf = np.sqrt(focus_p ** 2 + Rax ** 2 - 2 * focus_p * Rax * np.cos(alpha_rad))

Rfd = pd.DataFrame(Rf, index=ths, columns=[str(x) + "km" for x in height])
Rfd_norm = np.sqrt(1 ** 2 + focus_p ** 2 - np.cos(thax))

Rfd_power = (Rfd_norm ** 2 / Rfd ** 2)

cycle = uplt.Cycle(colors=plot_colors)

df_accu = pd.read_csv("./csv_highspeed/line_plot_distance_ratio.csv", index_col=0)
df_accu.columns = [str(x) + "km" for x in height]

fig, ax = uplt.subplots(suptitle="Rfd_power real focus point", xlabel="s/c angle alpha (degree)", ylabel="Rfd_power", figsize=(10,6))
ax.format(ylim=(0,1.0), ylocator=0.1)
ax.plot(ald, Rfd_power, cycle=cycle, label=[f"{h} km real focus" for h in height], ls="--", legend="ur")
ax.plot(df_accu_ax, df_accu, cycle=cycle, label=[f"{h} km accurate" for h in height], legend="ll")

fig.savefig(fp + "Rfd_power_reality_focus.png")
fig.show()
```

```python
focus_p = 1/2 * np.cos(thax)

theta_d = alpha_rad - thax

L = np.sqrt(Rax ** 2 + 1 ** 2 - 2 * 1 * Rax * np.cos(theta_d))

Rf = L + focus_p

Rfd = pd.DataFrame(Rf, index=ths, columns=[str(x) + "km" for x in height])
Rfd_norm = focus_p

Rfd_power = (Rfd_norm ** 2 / Rfd ** 2)

cycle = uplt.Cycle(colors=plot_colors)

#df_accu = pd.read_csv("./csv_highspeed/line_plot_distance_ratio.csv", index_col=0)
#df_accu.columns = [str(x) + "km" for x in height]

ymax = 1

fig, ax = uplt.subplots(suptitle="Rfd_power real focus point", xlabel="s/c angle alpha (degree)", ylabel="Rfd_power", figsize=(10,6))
ax.format(ylim=(0,ymax), ylocator=ymax/10)
ax.plot(ald, Rfd_power, cycle=cycle, label=[f"{h} km real focus" for h in height], ls="--", legend="ur")
ax.plot(df_accu_ax, df_accu, cycle=cycle, label=[f"{h} km accurate" for h in height], legend="ll")

fig.savefig(fp + "Rfd_power_reality_focus.png")
fig.show()
```

```python
focus_p = 1/2 * np.cos(thax)
focus_p2 = 1/(2 * np.cos(thax))

theta_d = alpha_rad - thax

L = np.sqrt(Rax ** 2 + 1 ** 2 - 2 * 1 * Rax * np.cos(theta_d))

Rf = L + focus_p
Rg = np.sqrt(focus_p2 ** 2 + 1 ** 2 -2 * focus_p2 * 1 * np.cos(thax))

Rfd = pd.DataFrame(Rf, index=ths, columns=[str(x) + "km" for x in height])
Rfd_norm = focus_p

gt = Rfd_norm / Rfd
gp = Rg / (L + Rg)

Rfd_power = gt * gp

cycle = uplt.Cycle(colors=plot_colors)

#df_accu = pd.read_csv("./csv_highspeed/line_plot_distance_ratio.csv", index_col=0)
#df_accu.columns = [str(x) + "km" for x in height]

ymax = 1

fig, ax = uplt.subplots(suptitle="Rfd_power real focus point", xlabel="s/c angle alpha (degree)", ylabel="Rfd_power", figsize=(10,6))
ax.format(ylim=(0,ymax), ylocator=ymax/10)
ax.plot(ald, Rfd_power, cycle=cycle, label=[f"{h} km real focus" for h in height], ls="--", legend="ur")
ax.plot(df_accu_ax, df_accu, cycle=cycle, label=[f"{h} km accurate" for h in height], legend="ll")

fig.savefig(fp + "Rfd_power_reality_focus.png")
fig.show()
```
